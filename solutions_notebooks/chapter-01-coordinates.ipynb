### Chapter Overview

This chapter introduces the **Cartesian coordinate system** — a powerful mathematical framework that uses two perpendicular number lines (axes) to locate points in a plane. You will learn:

- How to plot points using ordered pairs $(x, y)$
- The four quadrants and their sign conventions
- Distance of a point from the axes
- Real-world applications: room layouts, navigation, mapping

---

## 📐 Key Concepts & Formulas

### The Cartesian Plane
- **x-axis**: Horizontal number line
- **y-axis**: Vertical number line  
- **Origin O**: Point of intersection $(0, 0)$
- **Coordinates of a point P**: $(x, y)$ where
  - $x$ = perpendicular distance from y-axis (abscissa)
  - $y$ = perpendicular distance from x-axis (ordinate)

### Quadrant Sign Conventions
| Quadrant | x-coordinate | y-coordinate | Example |
|----------|--------------|--------------|---------|
| I        | +            | +            | $(3, 4)$ |
| II       | -            | +            | $(-3, 4)$ |
| III      | -            | -            | $(-3, -4)$ |
| IV       | +            | -            | $(3, -4)$ |

### Points on Axes
- On **x-axis**: $y = 0$ → $(x, 0)$
- On **y-axis**: $x = 0$ → $(0, y)$
- **Origin**: $(0, 0)$

### Distance from Axes
- Distance from **y-axis** = $|x|$ (absolute value of x-coordinate)
- Distance from **x-axis** = $|y|$ (absolute value of y-coordinate)

```python
# Python visualization helper for coordinate geometry
import matplotlib.pyplot as plt
import numpy as np

# Set up a clean plotting style
plt.style.use('seaborn-v0_8-whitegrid')

# Helper function to plot points and axes
def plot_coordinate_plane(points, labels=None, xlim=(-10, 10), ylim=(-10, 10), title="Cartesian Plane"):
    """Plot points on a Cartesian plane with axes and grid."""
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Draw axes
    ax.axhline(y=0, color='black', linewidth=1.5)
    ax.axvline(x=0, color='black', linewidth=1.5)
    
    # Set limits and aspect
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
    ax.set_aspect('equal')
    
    # Grid
    ax.grid(True, linestyle='--', alpha=0.5)
    
    # Labels
    ax.set_xlabel('x-axis', fontsize=12)
    ax.set_ylabel('y-axis', fontsize=12)
    ax.set_title(title, fontsize=14, fontweight='bold')
    
    # Plot points
    if points:
        xs = [p[0] for p in points]
        ys = [p[1] for p in points]
        ax.scatter(xs, ys, color='red', s=100, zorder=5)
        
        if labels:
            for i, (x, y) in enumerate(points):
                ax.annotate(f'{labels[i]}({x}, {y})', (x, y), 
                           xytext=(5, 5), textcoords='offset points',
                           fontsize=10, fontweight='bold')
    
    # Quadrant labels
    ax.text(xlim[1]*0.8, ylim[1]*0.8, 'I', fontsize=20, alpha=0.2, ha='center', va='center')
    ax.text(xlim[0]*0.8, ylim[1]*0.8, 'II', fontsize=20, alpha=0.2, ha='center', va='center')
    ax.text(xlim[0]*0.8, ylim[0]*0.8, 'III', fontsize=20, alpha=0.2, ha='center', va='center')
    ax.text(xlim[1]*0.8, ylim[0]*0.8, 'IV', fontsize=20, alpha=0.2, ha='center', va='center')
    
    plt.tight_layout()
    plt.show()

print("Coordinate geometry helper functions loaded.")
```

```python
# Solution to Practice Problem 3
# Given vertices
A = (2, 3)
B = (2, -1)
C = (6, -1)

# A and B share x=2 → vertical side AB
# B and C share y=-1 → horizontal side BC
# Fourth vertex D must have x from A and y from C
D = (A[0], C[1])
print(f"Fourth vertex D = {D}")

# Verify
print(f"AB vertical: x_A = x_B = {A[0]} ✓")
print(f"BC horizontal: y_B = y_C = {B[1]} ✓")
print(f"CD vertical: x_C = x_D = {C[0]} ✓")
print(f"DA horizontal: y_D = y_A = {D[1]} ✓")
```

### Problem 4
The point $M$ is the mirror image of $N(4, -3)$ in the $y$-axis. Find the coordinates of $M$.
